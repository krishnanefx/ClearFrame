## 4. Structured Output Analysis

Get results in a structured format using Pydantic models.

```python
from examples.valyu_holisticai_agent import create_structured_output_analyzer
import json

# Create structured analyzer
structured_analyzer = create_structured_output_analyzer()

# Sample analysis prompt
analysis_prompt = """
Analyze the following news comparison:

Topic: Climate Change Policy

Articles:
1. BBC News: "UK Sets New Climate Targets" 
   - Neutral tone (sentiment: 0.0)
   - Focuses on policy details and timelines
   - Mentions: UK Government, Paris Agreement, Net Zero

2. CNN: "Climate Crisis Deepens as Nations Fail to Act"
   - Negative tone (sentiment: -0.4)
   - Emphasizes urgency and failures
   - Mentions: UN Climate Report, Global Warming, Extreme Weather

Key differences:
- BBC frames as incremental progress
- CNN frames as crisis requiring urgent action
- BBC focuses on policy, CNN on impacts

Provide structured bias analysis.
"""

# Get structured result
result = structured_analyzer.invoke(analysis_prompt)

# Display results
if hasattr(result, 'model_dump'):
    print(json.dumps(result.model_dump(), indent=2))
else:
    print(json.dumps(result, indent=2))
```

## 5. Using Individual Tools

Access the analysis tools directly.

```python
from examples.valyu_holisticai_agent import (
    analyze_sentiment,
    extract_entities,
    calculate_bias_score
)

# Test sentiment analysis
text = "The new climate policy represents significant progress toward our environmental goals."
sentiment = analyze_sentiment.invoke({"text": text})
print("Sentiment Analysis:")
print(f"  Polarity: {sentiment['polarity']:.2f}")
print(f"  Classification: {sentiment['classification']}")
print()
```

```python
# Test entity extraction
text = "President Biden met with UN Secretary-General António Guterres in New York to discuss climate policy."
entities = extract_entities.invoke({"text": text})
print("Extracted Entities:")
for entity in entities:
    print(f"  - {entity}")
print()
```

```python
# Test bias score calculation
bias = calculate_bias_score.invoke({
    "state_aligned_count": 8,
    "independent_count": 5,
    "state_sentiment": 0.6,
    "independent_sentiment": 0.1,
    "omission_ratio": 0.4
})

print("Bias Metrics:")
print(f"  Coverage Delta: {bias['coverage_delta']:.2f}")
print(f"  Sentiment Delta: {bias['sentiment_delta']:.2f}")
print(f"  Omission Ratio: {bias['omission_ratio']:.2f}")
print(f"  Overall Bias Level: {bias['bias_level'].upper()}")
```

## 6. Complete Pipeline Example

Run a complete analysis from search to structured output.

```python
# Define topic and sources
topic = "artificial intelligence regulation"
sources = ["BBC News", "China Daily", "The Guardian"]

# Step 1: Search for articles
search_query = f"""
Search for recent articles about "{topic}" from these sources:
{', '.join(sources)}

For each article:
1. Extract title, source, and URL
2. Analyze sentiment
3. Extract key entities
4. Provide a brief summary
"""

print(f"Searching for: {topic}")
print(f"Sources: {', '.join(sources)}")
print("\n" + "="*60 + "\n")

search_response = agent.invoke({"input": search_query})
print(search_response['output'])
print("\n" + "="*60 + "\n")
```

```python
# Step 2: Analyze for bias
analysis_query = f"""
Based on the articles found about "{topic}", provide a comprehensive bias analysis.

Calculate:
- Coverage delta (difference in volume between state-aligned and independent sources)
- Sentiment delta (difference in tone)
- Omission ratio (what's left out)

Identify:
- Key narrative differences
- What each source emphasizes or omits
- Overall bias level (low/medium/high)

Provide recommendations for balanced information consumption.
"""

print("Analyzing bias...")
print("\n" + "="*60 + "\n")

analysis_response = agent.invoke({"input": analysis_query})
print(analysis_response['output'])
```

## 7. Generate Report

Create a formatted report from the analysis.

```python
from datetime import datetime

# Create a simple report
report = f"""
{'='*80}
NEWS BIAS ANALYSIS REPORT
{'='*80}

Topic: {topic}
Sources Analyzed: {', '.join(sources)}
Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}

{'='*80}
FINDINGS
{'='*80}

{analysis_response['output']}

{'='*80}
"""

print(report)

# Optionally save to file
# with open(f'report_{topic.replace(" ", "_")}.txt', 'w') as f:
#     f.write(report)
```
